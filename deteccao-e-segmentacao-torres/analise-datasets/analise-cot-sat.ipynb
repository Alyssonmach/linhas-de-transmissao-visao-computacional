### Importação dos Pacotes Utilitários

```python
from collections import Counter
import matplotlib.pyplot as plt
import json, os
```

### Funções Utilitárias
***

```python
def load_json_file(json_path):
    '''
    Carrega um arquivo json em um dicionário.

    Args:
        json_path (str) --> caminho relativo de acesso ao arquivo json no diretório.

    Returns:
        data (dict) --> dicionário contendo todas as informações do arquivo json.
    '''

    with open(json_path, 'r') as json_file:
        data = json.load(json_file)
    
    return data
```

```python
def initialize_images_ids_with_dict(data):
    '''
    Adquirindo os IDs das imagens a partir do dicionário com os rótulos.

    Args:
        data (dict) --> dicionário com as informações de rotulagem do conjunto de imagem.
    
    Returns:
        count_ids (dict) --> dicionário vazio com as chaves de IDs das imagens.
    '''

    count_ids = dict()
    for info in data['images']:
        count_ids[info['id']] = list()
    
    return count_ids
```

### Análise dos Dados
***

```python
# carregando os rótulos no arquivo json dentro de um dicionário
json_path1 = '../COT-SAT-dataset/train/_annotations.coco.json'
json_path2 = '../COT-SAT-dataset/valid/_annotations.coco.json'
json_path3 = '../COT-SAT-dataset/test/_annotations.coco.json'

data1 = load_json_file(json_path1)
data2 = load_json_file(json_path2)
data3 = load_json_file(json_path3)
```

```python
# analisando as chaves de acesso do dicionário
data1.keys()
```

```python
# analisando as informações do conjunto de imagens
data1['info']
```

```python
# analisando as categorias de rotulação dos dados com seus respectivos IDs
data1['categories']
```

```python
# inicializando um dicionário vazio com os IDs das imagens
images_ids1 = initialize_images_ids_with_dict(data1)
images_ids2 = initialize_images_ids_with_dict(data2)
images_ids3 = initialize_images_ids_with_dict(data3)

# obtendo a quantidade de instâncias de objeto por imagem
for info in data1['annotations']:
    images_ids1[info['image_id']].append(info['category_id'])

for info in data2['annotations']:
    images_ids2[info['image_id']].append(info['category_id'])

for info in data3['annotations']:
    images_ids3[info['image_id']].append(info['category_id'])
```

```python
# contando a quantidade de instâncias de objetos em todo o conjunto de imagens
verify_towers = {'Imagens com Torres': 0, 'Imagens sem Torres': 0}

for images_instances in images_ids1.values():
    count_aux = Counter(images_instances)
    if count_aux[1] != 0:
        verify_towers['Imagens com Torres'] += 1
    else:
        verify_towers['Imagens sem Torres'] += 1

for images_instances in images_ids2.values():
    count_aux = Counter(images_instances)
    if count_aux[1] != 0:
        verify_towers['Imagens com Torres'] += 1
    else:
        verify_towers['Imagens sem Torres'] += 1

for images_instances in images_ids3.values():
    count_aux = Counter(images_instances)
    if count_aux[1] != 0:
        verify_towers['Imagens com Torres'] += 1
    else:
        verify_towers['Imagens sem Torres'] += 1

print(verify_towers)
```
